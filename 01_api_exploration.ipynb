# Cell 2: a reusable helper

```python
def describe(obj, depth=0, max_depth=3):
    """Print the shape of a JSON object: keys, types and list sizes."""
    pad = "  " * depth
    if isinstance(obj, dict):
        for k, v in list(obj.items())[:15]:
            kind = type(v).__name__
            extra = f" (len {len(v)})" if isinstance(v, (list, dict)) else f" = {str(v)[:40]}"
            print(f"{pad}{k}: {kind}{extra}")
            if depth < max_depth and isinstance(v, (dict, list)):
                describe(v, depth + 1, max_depth)
    elif isinstance(obj, list) and obj:
        print(f"{pad}[0] ->")
        describe(obj[0], depth + 1, max_depth)
```
